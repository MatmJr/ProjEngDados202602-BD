```python
from pymongo import MongoClient
from pymongo.server_api import ServerApi
import os
from dotenv import load_dotenv
import certifi

load_dotenv()
```

```python
mongo_uri = os.getenv("MONGO_URI")

client = MongoClient(mongo_uri, server_api=ServerApi('1'), tlsCAFile=certifi.where())

db = client['IBGE']

print(db)
```

```python
data = db['PNADC'].find()
data
```

```python
serie = data[0]['resultados'][0]['series'][0]['serie']
serie
```

```python
import pandas as pd

df = pd.DataFrame.from_dict(serie, orient='index', columns=['valor'])
df.index.name = 'periodo'
df = df.reset_index()
df
```

```python
df.info()
```

```python
df['valor'] = df['valor'].replace('...', '0')
df['valor'] = df['valor'].astype(float)
```

```python
df.info()
```

```python
df['ano'] = df['periodo'].str[:4]
df['tri'] = df['periodo'].str[-1:]
df['periodo'] = pd.PeriodIndex(df['ano']+'Q'+df['tri'], freq='Q')
```

```python
df['periodo'] = df['periodo'].dt.to_timestamp()
df
```

```python
df.info()
```

```python
import sqlite3

conn = sqlite3.connect("IBGE.db")
df.to_sql("pnadc-total", con=conn, if_exists="replace", index=False)
conn.close()
```
